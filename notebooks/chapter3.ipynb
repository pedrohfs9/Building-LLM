```python
import torch

inputs = torch.tensor([
    [0.43, 0.15, 0.89], 
   [0.55, 0.87, 0.66], 
   [0.57, 0.85, 0.64], 
   [0.22, 0.58, 0.33], 
   [0.77, 0.25, 0.10],
   [0.05, 0.80, 0.55]
])
```

```python
query = inputs[1]

attn_scores_2 = torch.empty(inputs.shape[0])
for i, x_i in enumerate(inputs):
    attn_scores_2[i] = torch.dot(x_i, query)

print(attn_scores_2)
```

```python
def softmax_naive(x):
    return torch.exp(x) / torch.exp(x).sum(dim=0)

attention_weights_2_naive = softmax_naive(attn_scores_2)

print(f"Attention weights: {attention_weights_2_naive}")
print(f"Sum: {attention_weights_2_naive.sum()}")
```

```python
attn_weights_2 = torch.softmax(attn_scores_2, dim=0)

print(attn_weights_2)
```

```python
query = inputs[1]

context_vec_2 = torch.zeros(query.shape)
for i, x_i in enumerate(inputs):
    context_vec_2 += attn_weights_2[i]*x_i

print(context_vec_2)
```

```python
attn_scores = torch.empty(6, 6)

for i, x_i in enumerate(inputs):
    for j, x_j in enumerate(inputs):
        attn_scores[i, j] = torch.dot(x_i, x_j)

print(attn_scores)
```

```python
attn_scores = inputs @ inputs.T # multiplicação da matriz inicial pela matriz transposta (.T)

print(attn_scores)
```

```python
attn_weights = torch.softmax(attn_scores, dim=1)

print(attn_weights)
```

```python
sum_row_3 = sum([0.1385, 0.2379, 0.2333, 0.1240, 0.1082, 0.1581])

print(sum_row_3)
```

```python
all_context_vecs = attn_weights @ inputs

print(all_context_vecs)
```
