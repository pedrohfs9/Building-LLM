```python
import torch

inputs = torch.tensor([
    [0.43, 0.15, 0.89], 
   [0.55, 0.87, 0.66], 
   [0.57, 0.85, 0.64], 
   [0.22, 0.58, 0.33], 
   [0.77, 0.25, 0.10],
   [0.05, 0.80, 0.55]
])
```

```python
query = inputs[1]

attn_scores_2 = torch.empty(inputs.shape[0])
for i, x_i in enumerate(inputs):
    attn_scores_2[i] = torch.dot(x_i, query)

print(attn_scores_2)
```

```python
def softmax_naive(x):
    return torch.exp(x) / torch.exp(x).sum(dim=0)

attention_weights_2_naive = softmax_naive(attn_scores_2)

print(f"Attention weights: {attention_weights_2_naive}")
print(f"Sum: {attention_weights_2_naive.sum()}")
```

```python
attn_weights_2 = torch.softmax(attn_scores_2, dim=0)

print(attn_weights_2)
```

```python
query = inputs[1]

context_vec_2 = torch.zeros(query.shape)
for i, x_i in enumerate(inputs):
    context_vec_2 += attn_weights_2[i]*x_i

print(context_vec_2)
```

```python
attn_scores = torch.empty(6, 6)

for i, x_i in enumerate(inputs):
    for j, x_j in enumerate(inputs):
        attn_scores[i, j] = torch.dot(x_i, x_j)

print(attn_scores)
```

```python
attn_scores = inputs @ inputs.T # multiplicação da matriz inicial pela matriz transposta (.T)

print(attn_scores)
```

```python
attn_weights = torch.softmax(attn_scores, dim=1)

print(attn_weights)
```

```python
all_context_vecs = attn_weights @ inputs

print(all_context_vecs)
```

```python
context_length = attn_scores.shape[0]

mask_simple = torch.tril(torch.ones(context_length, context_length))
print(mask_simple)
```

```python
masked_simple = attn_weights*mask_simple
print(masked_simple)
```

```python
rows_sum = masked_simple.sum(dim=-1, keepdim=True)
masked_normalized = masked_simple / rows_sum
print(masked_normalized)
```

```python
mask = torch.triu(torch.ones(context_length, context_length), diagonal=1)
masked = attn_scores.masked_fill(mask.bool(), -torch.inf)
print(masked)
```

```python
# same result as before, using torch.inf -> more efficient

attn_weights = torch.softmax(masked / keys.shape[-1]**0.5, dim=-1)
print(attn_weights)
```
